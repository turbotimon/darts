```python
from darts.models.forecasting.nbeats import NBEATSModel
from darts.utils import timeseries_generation as tg
from darts import TimeSeries
```

```python
ts = tg.linear_timeseries(length=10)
past_cov = ts.copy() + 1

```

```python
model = NBEATSModel(
    input_chunk_length=5,
    output_chunk_length=2,
    n_epochs=1,
    batch_size=2,
    generic_architecture=False,
)
```

```python
model.fit(
    ts,
    past_covariates=past_cov,
)
```

```python
model.fit(
    ts.stack(past_cov),
)
```

## Conclusion

The implementation threads past_covars like multivariate series, but get rid of the output at the very end (end of the last stack)
